# Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import fetch_california_housing
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error, mean_squared_error,r2_score
import sqlite3

print("All libraries imported suuccessfully!")
```

# Load the DataSet

```python
# Load California Housing Dataset
housing =fetch_california_housing()

#Convert to DataFrame
df = pd.DataFrame(housing.data, columns = housing.feature_names)

#Add Target column (house price)
df["Price"] = housing.target

print("Dataset loaded successfully")
print("Shape",df.shape)
df.head()

```

```python
conn = sqlite3.connect("california_housing.db")
```

```python
df.to_sql("housing", conn, if_exists = "replace", index = False)
print("Data saved to the database")
```

```python
query = """
SELECT *
FROM housing
LIMIT 5;
"""
pd.read_sql_query(query,conn)
```

```python
query = '''
select count(*) As total_records
from housing;
'''
pd.read_sql_query(query,conn)
```

```python
query = '''
select AVG(Price) AS average_price
from housing;
'''
pd.read_sql_query(query,conn)
```

```python
query = '''
select MedInc, HouseAge, Price
from housing
where price > 3 AND HouseAge < 20
LIMIT 5;
'''
pd.read_sql_query(query, conn)
```

```python
query = '''
select count(*) AS missing_prices
from housing
where Price is NULL;
'''
pd.read_sql_query(query, conn)
```

```python
query = '''
select HouseAge, count(*) AS total_records,
AVG(Price) AS average_Price
from housing
Group By HouseAge
Order by HouseAge
limit 5;
'''
pd.read_sql_query(query, conn)
```
